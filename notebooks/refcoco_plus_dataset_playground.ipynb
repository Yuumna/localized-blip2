```python
from lavis.models import load_preprocess
from lavis.common.registry import registry
from omegaconf import OmegaConf

model_name = "blip2"
model_type = "coco"

model_cls = registry.get_model_class(model_name)
cfg = OmegaConf.load(model_cls.default_config_path(model_type)).preprocess
vis_processor, text_processor = load_preprocess(cfg)
```

```python
from datasets.refcoco_plus_dataset import RefCocoPlusDataset

MY_ROOT = "/project/dl2023s/hoos"

refcoco = RefCocoPlusDataset(
    vis_root=os.path.join(MY_ROOT, "shared/datasets/coco-2014/images"),
    instances_file=os.path.join(MY_ROOT, "shared/datasets/refcoco+/instances.json"),
    refs_file=os.path.join(MY_ROOT, "shared/datasets/refcoco+/refs(unc).p"),
    set_type="val",
    vis_transform=vis_processor["eval"],
)
```

```python
from torch.utils.data import DataLoader

dataloader = DataLoader(refcoco, batch_size=4, num_workers=1, collate_fn=RefCocoPlusDataset.collate_fn)
```
